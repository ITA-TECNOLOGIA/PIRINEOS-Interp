```python
from explainability.models import LLMConfig
from explainability.models import DefaultHFModelLoader, BnBQuantizedModelLoader
from explainability.tuned_lens import TunedLens
```

```python
# First example with no quantization
model_config = LLMConfig(
                model_name="Qwen/Qwen2.5-14B",
                device_map={"": 1},
                torch_dtype="auto",
            )
model_loader = DefaultHFModelLoader(model_config)   # This will prompt for HF token
model, tokenizer = model_loader.load_model()
tl = TunedLens(
    model=model,
    tokenizer=tokenizer,
    lens = "../tuned-lens-ITA/my_lenses/Qwen2/Qwen25-14B"
)
prompt_1 = "Betty is saving money for a new wallet which costs $100. Betty has only half of the money she needs. Her parents decided to give her $15 for that purpose, and her grandparents twice as much as her parents. How much more money does Betty need to buy the wallet?"
fig = tl.explain_output(
    prompt=prompt_1,
    statistic="entropy",
    layer_stride=2,
    token_range=(0, 1000),
    generation_kwargs={"max_new_tokens": 1000, "do_sample": True, "temperature": 0.8}
)
fig.show()
```
